```python
import pandas as pd
import numpy as np
from scipy.stats import kurtosis as k
a = pd.read_csv("programming_languages (1).csv")
a
```

```python
avg = a["popularity"]
kurto = k(avg, fisher=True)
print("Kurtosis:", kurto)
if kurto > 0:
    print("Leptokurtic")
elif kurto < 0:
    print("Platykurtic")
else:
    print("Mesokurtic")
```

```python
avg = a["popularity"]
kurto = k(avg, fisher=False)
print("Kurtosis:", kurto)
if kurto < 3:
    print("Platykurtic")
elif kurto > 3:
    print("Leptokurtic")
else:
    print("Mesokurtic")
```
