```python
import numpy as np
import pandas as pd
from sklearn.naive_bayes import MultinomialNB
```

```python
data=pd.read_csv("DailyDialog.csv")
data
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
text=data['text']
text
```

```python
vectorizer=CountVectorizer()
x=vectorizer.fit_transform(text)
```

```python
from sklearn.preprocessing import LabelEncoder
y=data['sentiment']
le=LabelEncoder()
y=le.fit_transform(y)
y
```

```python
model=MultinomialNB()
model.fit(x,y)
```

```python
sent=input("Enter your sentiment:")
ans=vectorizer.transform([sent])
```

```python
model.predict(ans)
```

```python
le.inverse_transform(model.predict(ans))
```
